#### data visualisation lib based on matplotlib provides high level interface for drawing attractive and informative statistical graphs.

```python
import seaborn as sns
```

```python
# basic plotting with seaborn 
tips=sns.load_dataset('tips')
tips
```

```python
# create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('plot 1')
plt.show()
```

```python
# line plot
sns.lineplot(x='size', y='total_bill', data=tips)
plt.title('total bill by size')
plt.show()
```

```python
# categorical plots

#bar plot
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('bar plot ot total bill by day')
plt.show()
```

```python
# box plot

sns.boxplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# voilion plot

sns.violinplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
# histograms

sns.histplot(tips['total_bill'], bins=10, kde=True)
```

```python
sns.histplot(tips['total_bill'], bins=10, kde=False)
```

```python
# Kde plot

sns.kdeplot(tips['total_bill'], fill=True)
```

```python
# Pair plot

sns.pairplot(tips)
```

```python
# heatmap

corr=tips[['total_bill', 'tip', 'size']].corr()
corr
```

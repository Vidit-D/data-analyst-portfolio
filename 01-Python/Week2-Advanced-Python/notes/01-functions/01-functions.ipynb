```python
## why function

num=24
if num%2==0:
    print('The number is even')
else:
    print('the number is odd')
```

```python
def even_or_odd(num):
    '''Hey this function finds even or odd'''
    if num%2==0:
        print('The number is even')
    else:
        print('The number is odd')
```

```python
# call this function
even_or_odd(24)
```

```python
# function with multiple parameters

def add(a,b):
    '''This fucntion add two numbers'''
    return a+b

result=add(2,4)
print(result)
```

```python
# default parameters

def greet(name):
    print(f'Hello {name}')

greet('vidit')
```

```python
# you will get an error beacuse we have not given any parameter
def greet(name):
    print(f'Hello {name}')

greet()
```

```python
# so now we will give a defaul value in case no value is given to us

def greet(name='Guest'):
    print(f'Hello {name}')

greet('Vidit')
greet()
```

```python
# variable length arguments
# positional and keyword arguments

def print_numbers(*vidit):  # you can use any word instead of vidit but it is common pracgice to use args
    for numbers in vidit:
        print(numbers)
```

```python
print_numbers(1,2,34,5,5,67,4.5, 'dalmia')
```

```python
# positional arguments
def print_numbers(*args):  # common practice to use args
    for numbers in vidit:
        print(numbers)
```

```python
# keywords arguments - in this all the parameters will in key value pair
# we use '**kwargs' commomly - kw means keyword and args means argument

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_details(name='vidit', age=32, country='india')
```

```python
# combinig both of them 

def print_details(*args, **kwargs):
    for val in args:
        print(f'Positional argument - {val}')

    for key,value in kwargs.items():
        print(f'keyword argument - {key}:{value}')
```

## function examples

```python
# example1 - temperature conversion

def convert_temperature(temp,unit):
    '''This function converts temperature between celcius and fahrenheit'''
    if unit=='C':
        return temp *9/5 +32 ## celcius to fahrenheit
    elif unit=='F':
        return (temp-32)*5/9 ## fahrenheit to celcius  
    else:
        return None

print(convert_temperature(25, 'C'))
print(convert_temperature(77, 'F'))
```

```python
# example2 - password strenght checker

def is_strong(password):
    '''This function checks if the password is trong or not'''
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True

#calling the function
print(is_strong('WeakPwd'))
print(is_strong('We2ashdg!'))

```

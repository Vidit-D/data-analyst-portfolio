# File Handling 
### **File operation read and write** - pythin has built in functions and methods to read and write to files both text and binary. 

```python
# read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
# read a file line by line 

with open('example.txt', 'r') as file:
    for line in file:
        print(line)
        
```

```python
# read a file line by line 

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  ## strip() removes the newline character
        
```

```python
## writing a file(Overwriting)

with open('example.txt', 'w') as file:
    file.write('Hello world!\n')  ##\n is used to create a new line
    file.write('This is a new line')
```

```python
## writing a file( without Overwriting)

with open ('example.txt' , 'a') as file:
    file.write('Append operation taking place!\n')
```

```python
## writing a list of lines to a file

lines=['this is my first line\n', 'thi is my second line\n', 'this is my third line\n']
with open('example.txt' ,'a') as file:
    file.writelines(lines)
```

```python
## Binary files - we write in bytes in binary files

# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# Reading a binary file
with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

```python
# read the content from a source text file and write it to a destination txt file

with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
# read a text file count the number of lines, chracters and words

def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, characters = count_text_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```

```python
# writing and then reading a file
'''w+ mode in python is used to opena  file both for reading an writing. 
if file dosent exist it will be created and if it exixts it will be overwritten'''

with open('example.txt', 'w+') as file:
    file.write('hello world!\n')
    file.write('this is a new line\n')

    # move the file curson to the first line to read 
    #file.seek(0)
    file.seek(0)

    #read the content of the file
    content = file.read()
    print(content)
```

```python
# checking if a path or directory exist

path = 'example.txt'

if os.path.isfile(path):
    print(f'The path {path} is a file')
elif os.path.isdir(path):
    print(f'The path {path} is a directory')
else:
    print(f'The path {path} is neither a file ore directory')
```

# Dictionaries
### unordered collection of items. Store data in key value paris. Keys must be unique and immutable



```python
# creating dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
print(type(empty_dict))
```

```python
student= {'name':'Vidit','age':23,'grade':24}
print(student)
print(type(student))
```

```python
# error
# if key is reused it gets repalced with the later one so always ise unique keys
student= {'name':'Vidit','age':23,'name':24}
print(student)
```

```python
# accessing dict elements

student= {'name':'Vidit','age':23,'grade':'A'}
print(student['grade'])
print(student['age'])

# accessing values using get() method
print(student.get('grade'))
# if it is not present in the dict it will show as none by default 
print(student.get('last_name'))

# how to get default value of a key which is not present
print(student.get('last_name', 'not available'))
```

```python
# modifying dict elements
# dict are mutable, so you can add, update or delete elements

print(student)
```

```python
student['age']=33 # updated the value of the key
print(student)
student['address']='India' # added a new key and value
print(student)
```

```python
# delete any key

del student['grade'] # will delete the key and value pair 
print(student)
```

```python
## dict methods

keys = student.keys() # get all the keys 
print(keys)
values = student.values() # get all the values
print(values)

items = student.items() # get all key value pairs
print(items)
```

```python
# shallow copy

student_copy=student
print(student)
print(student_copy)
```

```python
# here we see that both of the variables gets changed
student['name']='vidit1'
print(student)
print(student_copy)
```

```python
#for such reason we use shallow copy so what if we update it in one dict it dosnt get updated in the other
student_copy1=student.copy() # shallow copy

student['name']='vidit3'
print(student_copy1)
print(student)
```

```python
# iterating over dict
# we can use loops to iterate over dict, keys, values, items(key+values)

## iterating over keys
for keys in student.keys():
    print(keys)

## iterate over values
for value in student.values():
    print(value)

for item in student.items():
    print(item)

#or
for key,value in student.items():
    print({f'{key}:{value}'})
```

```python
# nested dict

students={
    'student1':{'name':'vidit', 'age':23},
    'student2':{'name':'peter', 'age':11}
}
print(students)
```

```python
# access nested dict elemets
print(students['student2']['name'])
print(students['student2']['age'])
```

```python
print(students)
```

```python
# iterating over nested dict

for student_id, student_info in students.items():
    print(f'{student_id}:{student_info}')
    for key, value in student_info.items():
        print(f'{key}:{value}')
```

```python
# dict comprehension

sqr={x:x**2 for x in range(5)} #here x is the key and x**2 is the value
print(sqr)
```

```python
## conditional dict comprehension

even_var= {x:x**2 for x in range(10) if x%2==0}
print(even_var)
```

```python
# practical examples

# use a dict to count the frequency of elements of list

numbers=[1,2,2,3,3,3,4,4,4,4]

frequency= {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
# merge two dict

dict1={'a':1, 'b':2}
dict2={'b':3, 'c':4}
merge_dict={**dict1, **dict2} # all the key value pairs of each will be combined
print(merge_dict)
```

# Sets
### collections of unique items. unordered but do not allow duplicate values

```python
# create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
# list to set

empty_set=set([1,2,3,4,5])
print(empty_set)
```

```python
my_set=set([1,2,3,4,5,4,6,7,6])
print(my_set)
```

```python
# basic set operations
# adding and removing elements
my_set.add(7)
print(my_set)
```

```python
# remove elements from the set
my_set.remove(3)
print(my_set)
```

```python
# since 10 was not present it will show an error
my_set.remove(10)
```

```python
# but what if we want to remove something if it is there and if its not it dosent give us an error

my_set.discard(11)
print(my_set)
```

```python
# pop method
removed_set=my_set.pop()
print(removed_set)
print(my_set)
```

```python
# clear all the elements
my_set.clear()
print(my_set)
```

```python
# Set membership test
# a test to see whether an element is part of that set or not 
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
# mathematical operations
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

# union
union_set= set1.union(set2)
print(union_set)

# intersection

intersection_set=set1.intersection(set2)
print(intersection_set)

# it will find out the common elements and update it for the values of set1
set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

# difference operation
# it removes the common element present in set1 and set2 and return the unique values in set1
x=set1.difference(set2)
print(x)
print(set2.difference(set1))
```

```python
# symmetric difference
# in this from both sets the common values are removed and it gives the unique ones
set1.symmetric_difference(set2)
```

```python
# Sets methods
set1={1,2,3,4,5}
set2={3,4,5}

# is subset
#Used to to check if all the elements present in set1 are also present in set2
print(set1.issubset(set2))

#superset
print(set1.issuperset(set2))
```

```python
lst=[1,2,2,3,4,4,5]

set(lst)
```

```python
## counting unique words in text

text='In this tutorial we are talking about sets'
words=text.split()

# convert list of words to set to get unique values 
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```

## Loops

```python
# for loop - used to iterate over a function

range(5)
```

```python
# last number is  not included as we start from 0
for i in range(5):
    print(i)
```

```python
for i in range(0,6):
    print(i)
```

```python
for i in range(1,10,1):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)

for j in range(10,1,-1):
    print(j)
```

```python
# strings

str='Vidit Dalmia'
for i in str:
    print(i)
```

```python
# while loop - continues to execute as long as the condition as true

count = 0

while count<5:
    print(count)
    count=count+1

```

```python
count = 0

while count%2==0:
    print(count)
    count=count+1

```

```python
# Loop control statement 

# break 
# - this statement exits the loop prematurely 

# break statement

for i in range(10):
    if i==5:
        break
    print(i)







```

```python
# continue statement

## the continue statement skips the current iteration and continues with the next 

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
#Pass statement 
# it is a null operation; it does nothing

for i in range(5):
    if i ==3:
        pass
    print(i)
```

```python
# Nested loops - a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f'i:{i} and j:{j}')
```

```python
# example - calculate the sum of first n natural numbers using a while and for loop

#while loop

n=10
sum=0
count=1

while count<=n:
    sum=sum+count
    count=count+1
print('Sum of first 10 natural numbers is:',sum)
```

```python
# for loop

n=10
sum=0

for i in range(11):
    sum = sum+i
print("The sum is:", sum)
```

```python
# examples - prime numbers between 1 and 100

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)

```

# Real world examples using lists in python


```python
#example - A to do list
# create a list to keep track of tasks

to_do_list=['Buy groceries','Clean the house', 'Pay bills']

#adding a tast
to_do_list.append('Schedule meeting')
to_do_list.append('Go for a run')

# removing a task

to_do_list.remove('Clean the house')

#checking if a task is in the list
if 'Pay bills' in to_do_list:
    print('Dont forget to pay the bills')

print('to do list remaining')
for task in to_do_list:
    print(f'-{task}')
```

```python
# Example2 - organising student grades
#create a list to store and calculte avg grades for students

# creating a list with student grades
grades=[78,44,90,67,89]

# adding a grade
grades.append(74)

#removing a grade
grades.remove(44)

# calculating the avg
avg_grade= sum(grades) / len(grades)
print(f'Avg grade is: {avg_grade}')

#finding lowest and highest grades
max_grade= max(grades)
min_grade = min(grades)
print(f'Highest grade is: {max_grade}')
print(f'lowest grade is: {min_grade}')
```

```python
# Example 3- Managing an inventory
# use a list to manage inventory items in a store 

# managing an inventory
inventory=['apples','bananas', 'guava', 'peach', 'orange']

# adding stock
inventory.append('grapes')

# removing stock
inventory.remove('peach')

# checking stock
item = 'bananas'
if item in inventory:
    print('Its in stock')
else:
    print('Its out of stock')

# printing the inventory 
print('Inventory list:')
for item in inventory:
    print(f'-{item}')
```

```python
# example 4- collectiong user feedback
# use a list to collect and analyse feedback

# Collecting user feedback
feedback = ["Great service!", "Very satisfied", "Could be better", "Excellent experience"]

# Adding new feedback
feedback.append("Not happy with the service")

# Counting specific feedback
positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())
print(f"Positive Feedback Count: {positive_feedback_count}")

# Printing all feedback
print("User Feedback:")
for comment in feedback:
    print(f"- {comment}")
```

## Conditional statements(if,else,elif)

```python
# if statement - executes the condition within itself

age=20

if age>=18:
    print('You are allowed inside the club')
```

```python
# else - it executes a block of code when the if statement is false

age=16

if age>=18:
    print('you are allowed inside the club')
else:
    print('You are a minor')
```

```python
# elif statemnt - allows you to check multiple statement 

age = 17

if age<13:
    print('You are a child')
elif age<18:
    print('you are a teenager')
else:
    print('You are an adult')
```

```python
## Nested conditional statements
# you can place one or more if,elif,else statement inside the if,else,elif statemnt 

# Tell me if a  number is odd,even,negative,positive

num=int(input('Enter the number'))

if num >0:
    print('The number is positive')
    if num%2==0:
        print('The number is even')
    else:
        print('The number is odd')
else:
    print("The number is zero or negative")
```

```python
# Use Cases/ Practical Examples

# Determine if a year is a leap year using nested conditional statement 

year=int(input('Plese enter a year'))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year, 'is a leap year')
        else:
            print(year, 'is not a leap year')
    else:
        print(year, 'is a leap year')
else:
    print(year, 'is not a leap year')
        
```

## Variables

```python
## Variables and Data Types - In Python, you can create variables to store data. Variables can hold different types of data, such as integers, floats, strings, lists, and more. The type of a variable is determined by the value assigned to it.

## Declaring and assigning variables
age=32
height=6.1
name="vidit"
is_student= True

print("age :", age)
print("height :", height)
print("name :", name)
print("is_student :", is_student)
```

```python
## Naming convention
## Variable names should be descriptive 
## They should alaways start with a letter or '_' and it can contain letters, numbers and underscores
## Variables are case sensitive

# valid variable names

first_name= 'Vidit'
last_name= 'Dalmia'
```

```python
# Invalid variables name
# 2age= 30
# first-name='vidit'
# name='vidit'
```

```python
# Case Sensitivity 
name='vidit'
Name='Dalmia'
print(name)
print(Name)
```

```python
## Understanding variable types
## Python is dynamically types, the type of the variable is determined at runtime 
age=25 #int
height=6.1 #float
name='vidit' # string
is_student= True #bool

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

```python
## Dynamic typing 
## Python allows the type of variable to change as the program executes 
var=10
print(var, type(var))

var='Hello'
print(var, type(var))

var=3.10
print(var, type(var))
```

```python
## Input - when the user gives some values

age=input('What is your age?')
print(age, type(age))

age2=int(input('What is your age?'))
print(age2, type(age2))
```

```python
## Simple calculator

num1= float(input('Enter number 1'))
num2= float(input('Enter number 2'))

sum = num1 + num2
diff = num1 - num2
multiply = num1* num2
divide = num1 / num2

print('The sum is :', sum)
print('The diff is :', diff)
print('The multiply is :', multiply)
print('The divide is :', divide)
```

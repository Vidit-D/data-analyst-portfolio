```python
year = int(input('Please enter the year'))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print('It is not a leap year')
        else:
            print('It is not a leap year')
    else:
        print('It is a leap year')
else:
    print('It is not a leap year ')
```

```python
num1 = float(input('Please enter a number'))
num2 = float(input('Please enter a number'))
opr = input('Please enter the operation you want to perform: +, -, *, /')
result = 0

if opr == '+':
    result = num1+num2
elif opr == '-':
    result = num1-num2
elif opr == '*':
    result = num1*num2
elif opr == '/':
    if num2!=0:
        result = num1/num2
    else:
        print('Invalid operation')
else:
    print('Invalid operator')

print('The result is: ',result)
```

```python
# calculate the cost of items in a cart 


def cart_items(cart):
    total_cost=0
    for item in cart:
        total_cost += item['cost']*item['quantity']
    return total_cost


cart= [
    {'name':'orange', 'cost':43, 'quantity':4}, 172
    {'name':'banana', 'cost':38, 'quantity':3}, 114
    {'name':'apple', 'cost':48, 'quantity':5} 240
]

#calling the function
total_cost=cart_items(cart)
print(total_cost)
```

```python
# a string is a palindrome or not

def is_palin(s):
    s=s.lower().replace(' ','')
    return s==s[::-1]

print(is_palin('A man a plan a canal panama'))
```

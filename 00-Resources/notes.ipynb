## Conditional statements(if,else,elif)

```python
# if statement - executes the condition within itself

age=20

if age>=18:
    print('You are allowed inside the club')
```

```python
# else - it executes a block of code when the if statement is false

age=16

if age>=18:
    print('you are allowed inside the club')
else:
    print('You are a minor')
```

```python
# elif statemnt - allows you to check multiple statement 

age = 17

if age<13:
    print('You are a child')
elif age<18:
    print('you are a teenager')
else:
    print('You are an adult')
```

```python
## Nested conditional statements
# you can place one or more if,elif,else statement inside the if,else,elif statemnt 

# Tell me if a  number is odd,even,negative,positive

num=int(input('Enter the number'))

if num >0:
    print('The number is positive')
    if num%2==0:
        print('The number is even')
    else:
        print('The number is odd')
else:
    print("The number is zero or negative")
```

```python
# Use Cases/ Practical Examples

# Determine if a year is a leap year using nested conditional statement 

year=int(input('Plese enter a year'))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year, 'is a leap year')
        else:
            print(year, 'is not a leap year')
    else:
        print(year, 'is a leap year')
else:
    print(year, 'is not a leap year')
        
```

```python
# assignment 
# simple calculator program

num1 = float(input('Please enter num1'))
num2 = float(input('Please enter num2'))
opr = input('Please enter the operation +,-,*,/')

if opr== '+':
    print(num1 + num2)
elif opr == '-':
    print(num1 - num2)
elif opr == '*':
    print(num1 * num2)
elif opr == '/':
    print(num1 / num2)
    if num2 !=0:
        print(num1 / num2)
    else:
        print("Invalid operation")
else:
    print('Please enter one of the given operators')

```

### List comprehension

Basic syntax [expression for the item in iterable]

with conditional logic [expression for the item in iterable if condition]

nested list comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
## basic list comprehension

square=[num**2 for num in range(10)]
print(square)
```

```python
### List comprehension with condition

lst=[]

for i in range(10):
    if i%2==0:
        lst.append(i)
print(lst)
        




```

```python
# now useing list comprehension

even_num=[num for num in range(10) if num%2==0]
print(even_num)
```

```python
# nested list comprehension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j]for i in lst1 for j in lst2]
print(pair)
```

```python
# list comprehension with function calls

words=['vidit', 'ambika', 'dalmia', 'aditya', 'paras']

length = [len(word) for word in words]
print(length)
```
